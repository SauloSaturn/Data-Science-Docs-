## <font color = lightblue> 1.1. Vetores



<font color = lightblue> **- Relação Entre Vetores e Dados**



**A Unidade Fundamental dos Dados**: Um vetor é uma entidade matemática que carrega a informação de **magnitude** e **direção**. Exatamente por esses dois motivos que os vetores são a melhor ferramenta para representarem os dados, além do que eles ostentarem o título de uma das entidades mais poderosas e versáteis dentro da matemática. Sendo assim, tratando dados como vetores existe um arsenal robusto para lidar com eles nas mais inusitadas situações e circunstâncias. Por isso a importância de saber álgebra linear, pois essa é a área da matemática que estuda natureza dos vetores.

Concluindo:  uma única observação ou entidade de um conjunto de dados, composta por múltiplos atributos (ou **features**), é naturalmente representada como um vetor em um espaço $n$-dimensional $(\mathbb{R}^n)$. Cada componente do vetor corresponde ao valor de um atributo específico. Por exemplo, um cliente pode ser descrito por um vetor contendo sua idade, renda e histórico de compras.

<font color = lightblue> **Exemplo 1**:

**Analogia do mundo real**: Um sujeito $A$ pensa em dar uma **instrução** a um sujeito $B$ no centro de uma cidade.
- Se o sujeito $A$ diz "ande $500$ metros", a informação está incompleta. $500$ metros para onde?
- Se o sujeito $A$ diz "ande $500$ metros na direção norte", essa **instrução** é um **vetor**! Ela tem:
 - Magnitude: $500$ metros
 - Direção: Norte

**Analogia em dados**: Em ciência de dados, um vetor pode representar um único cliente, um paciente ou uma casa. Por exemplo, um vetor **casa_A** poderia ser **[120, 3, 2]**, onde cada número no vetor representa uma característica:

- $120 m²$ (magnitude da área)
- $3$ quartos (magnitude dos quartos)
- $2$ banheiros (magnitude dos banheiros)


<font color = lightblue> **- Definição 1: Vetores**


Um vetor $\vec{v}$ no espaço euclidiano de $n$ dimensões, denotado por $\mathbb{R}^n$, é uma ênupla (ou tupla) ordenada de $n$ números reais. Escrevemos $\vec{v} = (v_1, v_2, \dots, v_n)$, onde cada $v_i \in \mathbb{R}$ é chamado de **componente** ou **coordenada** do vetor.

* O espaço $\mathbb{R}^2$ é o plano cartesiano já conhecido desde a escola.
* O espaço $\mathbb{R}^3$ é o espaço tridimensional.

**Notação:**
Em Álgebra Linear é mais comum e formal representar vetores como **matrizes coluna** (um caso particular de matriz):


$$\vec{v} = \begin{bmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{bmatrix}$$

<font color = lightblue> **- Interpretação Geométrica:**

A maneira mais intuitiva de pensar em um vetor é como um **segmento de reta orientado** — uma "seta" — que parte da origem do sistema de coordenadas e aponta para o ponto. Assim, todo vetor possui duas características fundamentais: **magnitude** (seu comprimento) e **direção**.

###<font color = lightblue> 1.1.1. Adição de Multiplicação por Escalar

<font color = lightblue> **- Definição 2: Adição**

Sejam os vetores $\vec{u} = (u_1, u_2, \dots, u_n)$ e $\vec{v} = (v_1, v_2, \dots, v_n)$ pertencentes ao espaço $\mathbb{R}^n$. A **soma** $\vec{u} + \vec{v}$ é o vetor resultante da soma de seus componentes correspondentes:

$$\vec{u} + \vec{v} = (u_1 + v_1, u_2 + v_2, \dots, u_n + v_n)$$

<font color = lightblue> **Exemplo 2:**

Se $\vec{u} = \begin{bmatrix} 1 \\ -2 \\ 3 \end{bmatrix}$ e $\vec{v} = \begin{bmatrix} 4 \\ 5 \\ -1 \end{bmatrix}$ em $\mathbb{R}^3$, a soma é:
$$\vec{u} + \vec{v} = \begin{bmatrix} 1+4 \\ -2+5 \\ 3+(-1) \end{bmatrix} = \begin{bmatrix} 5 \\ 3 \\ 2 \end{bmatrix}$$

<font color = lightblue> **- Propriedades Formais da Adição:**

Para quaisquer vetores $\vec{u}, \vec{v}, \vec{w} \in \mathbb{R}^n$, a adição satisfaz as seguintes propriedades:
-   **Comutatividade:** $\vec{u} + \vec{v} = \vec{v} + \vec{u}$
-  **Associatividade:** $(\vec{u} + \vec{v}) + \vec{w} = \vec{u} + (\vec{v} + \vec{w})$
-  **Existência do Elemento Neutro:** Existe um **vetor nulo** $\vec{0} = (0, \dots, 0)$ tal que $\vec{v} + \vec{0} = \vec{v}$.
-  **Existência do Elemento Oposto:** Para todo vetor $\vec{v}$, existe um vetor oposto $-\vec{v} = (-v_1, \dots, -v_n)$ tal que $\vec{v} + (-\vec{v}) = \vec{0}$.

<font color = lightblue> **- Interpretação Geométrica:**

A soma $\vec{u}+\vec{v}$ corresponde à diagonal do paralelogramo formado pelos vetores $\vec{u}$ e $\vec{v}$ (Regra do Paralelogramo), ou ao caminho direto do início de $\vec{u}$ ao fim de $\vec{v}$ se colocarmos a "cauda" de $\vec{v}$ na "cabeça" de $\vec{u}$ (Regra da Poligonal ou "cabeça na cauda").


<font color = lightblue> **- Definição 3: Multiplicação por Escalar:**

Seja $\vec{v} = (v_1, \dots, v_n)$ um vetor em $\mathbb{R}^n$ e $c \in \mathbb{R}$ um número real (chamado de **escalar**). O produto $c\vec{v}$ é o vetor obtido pela multiplicação de cada componente de $\vec{v}$ por $c$:

$$c\vec{v} = (cv_1, cv_2, \dots, cv_n)$$

<font color = lightblue> **Exemplo 3:**

Se $\vec{v} = \begin{bmatrix} 2 \\ 5 \\ -1 \end{bmatrix}$ e o escalar $c=3$, o produto é:
$$3\vec{v} = \begin{bmatrix} 3 \cdot 2 \\ 3 \cdot 5 \\ 3 \cdot (-1) \end{bmatrix} = \begin{bmatrix} 6 \\ 15 \\ -3 \end{bmatrix}$$

<font color = lightblue> **- Propriedades Formais da Multiplicação por Escalar:**

Para quaisquer vetores $\vec{u}, \vec{v} \in \mathbb{R}^n$ e escalares $c, d \in \mathbb{R}$:
-  **Associatividade de Escalares:** $c(d\vec{v}) = (cd)\vec{v}$
-  **Distributividade sobre a Soma de Vetores:** $c(\vec{u} + \vec{v}) = c\vec{u} + c\vec{v}$
-  **Distributividade sobre a Soma de Escalares:** $(c+d)\vec{v} = c\vec{v} + d\vec{v}$
-  **Existência do Elemento Neutro da Multiplicação:** O escalar $1$ atua como identidade: $1\vec{v} = \vec{v}$.

<font color = lightblue> **- Interpretação Geométrica:**

Multiplicar um vetor por um escalar $c$ "escala" o vetor. Se $|c|>1$, o vetor é esticado. Se $0 < |c| < 1$, o vetor é encolhido. Se $c<0$, o sentido do vetor é invertido. A direção original (a reta sobre a qual o vetor repousa) é preservada.

**(Observação Importante):** Estas oito propriedades que listamos (quatro da adição e quatro da multiplicação por escalar) são os **axiomas** que definem formalmente uma estrutura matemática chamada **Espaço Vetorial**. O fato de $\mathbb{R}^n$ satisfazê-las é o que nos permite aplicar toda a teoria da Álgebra Linear.

###<font color = lightblue> 1.1.2. Norma, Magnitude ou Comprimento de um Vetor

Até agora, foi definido que um vetor e suas interações com outros vetores (soma) ou com escalares (multiplicação). Mas como quantificar a noção intuitiva de "comprimento" de uma seta? A resposta está no conceito de **norma**.

<font color = lightblue> **- Definição 4: Norma Euclidiana**

A norma de um vetor $\vec{v} = (v_1, v_2, \dots, v_n)$ em $\mathbb{R}^n$, denotada por $||\vec{v}||$, é um escalar não-negativo que representa seu comprimento. Ela é calculada pela raiz quadrada da soma dos quadrados de seus componentes.

$$||\vec{v}|| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2} = \left(\sum_{i=1}^{n} v_i^2\right)^{1/2}$$

**Conexão Fundamental:** Esta fórmula nada mais é do que o **Teorema de Pitágoras** generalizado para um espaço de $n$ dimensões!

<font color = lightblue> **Exemplo 4:**

Vamos calcular a norma do vetor

$$\vec{v} = \begin{bmatrix} 3 \\ 4 \end{bmatrix}$$

em $\mathbb{R}^2$.

$$||\vec{v}|| = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5$$

O comprimento da seta que vai da origem ao ponto (3,4) é exatamente 5 unidades.

<font color = lightblue> **- Propriedades Formais da Norma:**

Para qualquer vetor $\vec{v}, \vec{u} \in \mathbb{R}^n$ e qualquer escalar $c \in \mathbb{R}$:
-  **Não-negatividade (Positividade Definida):** A norma é sempre positiva ou nula. Ela é zero se, e somente se, o vetor for o vetor nulo. $||\vec{v}|| \ge 0$, e $||\vec{v}|| = 0 \iff \vec{v} = \vec{0}$.

-  **Homogeneidade Absoluta:** Multiplicar um vetor por um escalar $c$ escala sua norma pelo valor *absoluto* de $c$. $||c\vec{v}|| = |c| \cdot ||\vec{v}||$.
-  **Desigualdade Triangular:** A norma da soma de dois vetores nunca é maior que a soma de suas normas. $||\vec{u} + \vec{v}|| \le ||\vec{u}|| + ||\vec{v}||$.(A intuição geométrica é que "um atalho nunca é mais longo do que o caminho original". O caminho direto, $\vec{u}+\vec{v}$, é sempre menor ou igual à soma dos comprimentos dos caminhos individuais, $||\vec{u}|| + ||\vec{v}||$).

**Conceito Derivado: Vetor Unitário**
Um vetor cuja norma é igual a $1$ é chamado de **vetor unitário**. Vetores unitários são extremamente importantes porque eles carregam apenas a informação de **direção**.

É possível transformar qualquer vetor não-nulo $\vec{v}$ em um vetor unitário (que denotamos com um acento circunflexo, $\hat{v}$) que aponta na mesma direção. Este processo é chamado de **normalização** e é feito dividindo o vetor pela sua própria norma:

$$\hat{v} = \frac{\vec{v}}{||\vec{v}||}$$

Este conceito de norma é o que nos permite falar em distância e comprimento no nosso espaço vetorial.

Quando estiver pronto e com o conceito de norma documentado, me avise para avançarmos para a última grande operação com vetores: o **Produto Escalar**. É ele que vai unir a ideia de norma com a de ângulo e nos abrir as portas para as aplicações mais interessantes.

### <font color = lightblue> 1.1.3. Produto Escalar (ou Produto Interno Euclidiano)

Introduzimos operações que resultam em vetores (soma, multiplicação por escalar). O produto escalar é diferente: ele é uma operação que recebe dois vetores e retorna um único **escalar** (um número). Sua verdadeira utilidade está no que este número nos diz sobre a relação geométrica entre os vetores.

<font color = lightblue> **Definição 5: Produto Escalar**

Sejam os vetores $\vec{u} = (u_1, \dots, u_n)$ e $\vec{v} = (v_1, \dots, v_n)$ em $\mathbb{R}^n$. O **produto escalar** de $\vec{u}$ e $\vec{v}$, denotado por $\vec{u} \cdot \vec{v}$, é o escalar obtido pela soma dos produtos dos seus componentes correspondentes:

$$\vec{u} \cdot \vec{v} = u_1v_1 + u_2v_2 + \dots + u_nv_n = \sum_{i=1}^{n} u_iv_i$$

<font color = lightblue> **Exemplo 5:**

Sejam $\vec{u} = \begin{bmatrix} 1 \\ 2 \\ -3 \end{bmatrix}$ e $\vec{v} = \begin{bmatrix} 4 \\ 0 \\ 2 \end{bmatrix}$ em $\mathbb{R}^3$.

$$\vec{u} \cdot \vec{v} = (1)(4) + (2)(0) + (-3)(2) = 4 + 0 - 6 = -2$$

Note que o resultado é o escalar $-2$, e não um vetor.

<font color = lightblue> **Propriedades Formais do Produto Escalar:**

Para quaisquer vetores $\vec{u}, \vec{v}, \vec{w} \in \mathbb{R}^n$ e um escalar $c \in \mathbb{R}$:
-  **Comutatividade:** $\vec{u} \cdot \vec{v} = \vec{v} \cdot \vec{u}$
-  **Distributividade sobre a Adição:** $\vec{u} \cdot (\vec{v} + \vec{w}) = (\vec{u} \cdot \vec{v}) + (\vec{u} \cdot \vec{w})$
-  **Homogeneidade (Associatividade com Escalar):** $(c\vec{u}) \cdot \vec{v} = c(\vec{u} \cdot \vec{v})$
-  **Positividade Definida:** $\vec{v} \cdot \vec{v} \ge 0$, e $\vec{v} \cdot \vec{v} = 0 \iff \vec{v} = \vec{0}$.

<font color = lightblue> **Teoremas e a Conexão Geométrica**

<font color = lightblue> **Teorema 1: Relação entre Produto Escalar e Norma**

O produto escalar de um vetor por si mesmo é o quadrado de sua norma.
$$\vec{v} \cdot \vec{v} = v_1^2 + \dots + v_n^2 = ||\vec{v}||^2$$
Este teorema é a ponte crucial entre as duas definições.

<font color = lightblue>**Teorema 2: A Fórmula Geométrica do Produto Escalar**

O produto escalar pode também ser expresso em termos das normas dos vetores e do ângulo $\theta$ entre eles:

$$\vec{u} \cdot \vec{v} = ||\vec{u}|| \cdot ||\vec{v}|| \cdot \cos(\theta)$$

Esta fórmula é a razão pela qual o produto escalar é tão útil. Ao igualar as duas formas de calculá-lo (a algébrica e a geométrica), podemos encontrar o ângulo entre dois vetores!

<font color = lightblue> **Definição 6: Ângulo entre Vetores**

A partir do **Teorema 2**, podemos definir o ângulo $\theta$ entre dois vetores não-nulos $\vec{u}$ e $\vec{v}$ como:

$$\cos(\theta) = \frac{\vec{u} \cdot \vec{v}}{||\vec{u}|| \cdot ||\vec{v}||}$$

<font color = lightblue> **Corolário 1: A Condição de Ortogonalidade**

A consequência mais importante disso tudo é a forma de determinar se dois vetores são perpendiculares (ou **ortogonais**). Dois vetores são ortogonais quando o ângulo entre eles é $\theta = 90^\circ$. Como $\cos(90^\circ) = 0$, a fórmula acima se simplifica para:

- **Dois vetores não-nulos $\vec{u}$ e $\vec{v}$ são ortogonais se, e somente se, o seu produto escalar for zero: $\vec{u} \cdot \vec{v} = 0$.**

Isso é uma ferramenta de imenso poder na Álgebra Linear e em todas as suas aplicações.

Ótima ideia. Vamos fazer exatamente isso. Um exemplo numérico é a melhor forma de ver como a teoria funciona na prática e como as fórmulas se conectam.

Vamos usar um exemplo para ilustrar as duas ideias ao mesmo tempo, pois elas são inseparáveis. Para encontrar o ângulo, precisamos usar a fórmula geométrica.

Este será o nosso **exemplo prático completo** para fechar o tópico de vetores.

<font color= lightblue> **Exemplo 6**:

Vamos encontrar o ângulo $\theta$ entre os vetores de $\mathbb{R}^2$:
$\vec{u} = \begin{bmatrix} 3 \\ 0 \end{bmatrix}$ e $\vec{v} = \begin{bmatrix} 2 \\ 2 \end{bmatrix}$.

Geometricamente, você pode imaginar o vetor $\vec{u}$ como uma seta de comprimento 3 totalmente sobre o eixo $X$, e o vetor $\vec{v}$ como uma seta que aponta "para cima e para a direita". Esperamos um ângulo agudo $(<90^\circ)$.

- Parte 1: Resolução Analítica**

A nossa ferramenta principal é a **Definição 6 (Ângulo entre Vetores)**, que é um rearranjo da fórmula geométrica do produto escalar:

$$\cos(\theta) = \frac{\vec{u} \cdot \vec{v}}{||\vec{u}|| \cdot ||\vec{v}||}$$

Para resolver, precisamos calcular os três componentes desta equação: o produto escalar e as duas normas.

**Passo 1: Calcular o Produto Escalar (o numerador)**
Usando a **Definição 5**:

$$\vec{u} \cdot \vec{v} = (u_1v_1) + (u_2v_2) = (3)(2) + (0)(2) = 6 + 0 = 6$$

Como o resultado é positivo, já confirmamos nossa intuição de que o ângulo é agudo.

**Passo 2: Calcular a Norma de $\vec{u}$ (parte do denominador)**
Usando a **Definição 4**:

$$||\vec{u}|| = \sqrt{3^2 + 0^2} = \sqrt{9} = 3$$

**Passo 3: Calcular a Norma de $\vec{v}$ (outra parte do denominador)**
Usando a **Definição 4**:

$$||\vec{v}|| = \sqrt{2^2 + 2^2} = \sqrt{4 + 4} = \sqrt{8}$$

Podemos simplificar $\sqrt{8} = \sqrt{4 \cdot 2} = 2\sqrt{2}$.

**Passo 4: Montar a equação e encontrar $\cos(\theta)$**
Agora substituímos os valores que encontramos na fórmula:

$$\cos(\theta) = \frac{6}{3 \cdot \sqrt{8}} = \frac{6}{3 \cdot 2\sqrt{2}} = \frac{6}{6\sqrt{2}} = \frac{1}{\sqrt{2}}$$

Para racionalizar o denominador (multiplicando em cima e embaixo por $\sqrt{2}$), obtemos:$$\cos(\theta) = \frac{\sqrt{2}}{2}$$

**Passo 5: Encontrar o Ângulo $\theta$**
Finalmente, encontramos o ângulo cujo cosseno é $\frac{\sqrt{2}}{2}$:

$$\theta = \arccos\left(\frac{\sqrt{2}}{2}\right) = 45^\circ$$

Nossa intuição estava correta, e agora temos o valor exato.

- **Parte 2: Resolução Numérica (Com Python)**

Vamos confirmar nosso resultado usando `NumPy`. O código abaixo reflete exatamente os passos que fizemos analiticamente.


```python
import numpy as np

# 1. Definir os vetores como arrays do numpy
u = np.array([3, 0])
v = np.array([2, 2])

# 2. Calcular o produto escalar
# Outra forma de escrever seria: produto_escalar = np.dot(u, v)
produto_escalar = u @ v

# 3. Calcular a norma (comprimento) de cada vetor
norma_u = np.linalg.norm(u)
norma_v = np.linalg.norm(v)

# 4. Calcular o cosseno do ângulo
# Adicionamos uma verificação para evitar divisão por zero se um dos vetores for nulo
if norma_u == 0 or norma_v == 0:
    cos_theta = 0
else:
    cos_theta = produto_escalar / (norma_u * norma_v)

# 5. Calcular o ângulo em radianos usando o arco-cosseno
# np.clip garante que o valor esteja entre -1 e 1 para evitar erros de ponto flutuante
angulo_radianos = np.arccos(np.clip(cos_theta, -1.0, 1.0))

# 6. Converter o ângulo para graus para melhor interpretação
angulo_graus = np.degrees(angulo_radianos)

# Exibir os resultados
print(f"Vetor u: {u}")
print(f"Vetor v: {v}")
print("-" * 25)
print(f"Produto Escalar (u . v): {produto_escalar}")
print(f"Norma de u (||u||): {norma_u:.4f}")
print(f"Norma de v (||v||): {norma_v:.4f}")
print("-" * 25)
print(f"Ângulo em Radianos: {angulo_radianos:.4f}")
print(f"Ângulo em Graus: {angulo_graus:.2f}°")
```
